```python
import requests

def download_image(num, dir="./images/{}.jpeg"):

    API = "https://api.rand.by/image"
    response = requests.get(API)
    if response.status_code==200:
        image_address = response.json()["urls"]["small"][:-3]+"512"
        with open(dir.format(num), "wb") as f:
            response = requests.get(image_address)
            if response.status_code==200:
                f.write(response.content)
```

```python
for i in range(12, 21):
    download_image(i, "./train/{}.jpeg")
```
